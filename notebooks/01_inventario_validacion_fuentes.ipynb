## Criterios de selección del video de referencia

Para el MVP se seleccionará inicialmente un video por género. Los 10
videos resultantes formarán el conjunto exploratorio de referencia y
cada uno deberá cumplir:

- Una sola persona.
- Cuerpo completo visible.
- Cámara frontal fija (`c01`).
- Video refinado, sin pre-roll ni post-roll.
- Situación `sBM` (Basic Dance).
- Resolución y duración adecuadas para el análisis.

La situación `sGR` (Group Dance) no se utilizará en el MVP porque implica múltiples personas y la estrategia actual delimita el sistema a una sola persona. Su análisis queda reservado para una futura versión.

La selección se basa en el alcance definido del MVP y en los metadatos publicados por AIST, no en una afirmación de que `sBM` sea una categoría superior para evaluar danza.

## Decisión de la fuente de referencia

Se utilizarán 10 videos refinados de AIST: uno por cada género disponible
en el subconjunto seleccionado. Todos cumplirán las siguientes condiciones:

- Situación: `sBM` (Basic Dance).
- Cámara: `c01` (cámara frontal fija).
- Una persona identificada mediante el campo `DANCER`.
- Video refinado, sin pre-roll ni post-roll.
- Calidad de 2 Mbps para reducir el tamaño de descarga durante el MVP.
- Duración suficiente para probar extracción de pose y comparación.

La selección de `sBM` responde al alcance del MVP, que inicialmente
analiza a una sola persona. Los videos `sGR` se reservan para una futura
versión con detección de múltiples personas. La selección de un registro
por género permite probar diversidad de estilos sin descargar todavía
todo el catálogo.

## Regla reproducible de selección

A partir del subconjunto `sBM` con cámara `c01`, se ordenan los registros
por `FILE_NAME` y se conserva el primer registro de cada género. En este
catálogo, la regla selecciona de forma determinista la coreografía `ch01`.
Esto no significa que `ch01` sea superior; únicamente evita una selección
aleatoria y permite reproducir exactamente el conjunto del MVP.

El resultado esperado es un video por género, diez videos en total.

```python
candidates_by_genre = {}

for row in sorted(reference_source_rows, key=lambda item: item["FILE_NAME"]):
    genre = row["GENRE"]

    if genre not in candidates_by_genre:
        candidates_by_genre[genre] = {
            "file_name": row["FILE_NAME"],
            "url": row["URL"],
            "genre": row["GENRE"],
            "situation": row["SITUATION"],
            "camera": row["CAMERA"],
            "dancer": row["DANCER"],
            "duration_sec": float(row["SEC"]),
            "file_size_bytes": int(row["FILE_SIZE"]),
        }

for candidate in candidates_by_genre.values():
    print(candidate)
    
```

```python
total_bytes = sum(candidate["file_size_bytes"] for candidate in candidates_by_genre.values())

total_mb = total_bytes / (1024 ** 2)

print(f"Videos candidatos: {len(candidates_by_genre)}")
print(f"Tamaño total aproximado: {total_mb:.2f} MB")
```

```python
from urllib.request import urlopen
from shutil import copyfileobj

download_results = []

for candidate in candidates_by_genre.values():
    target_path = REFERENCE_DIR / candidate["file_name"]

    if target_path.exists():
        download_results.append({
            "file_name": candidate["file_name"],
            "status": "already_exists",
            "size_bytes": target_path.stat().st_size,
        })
        continue

    print(f"Descargando: {candidate['file_name']}")

    try:
        with urlopen(candidate["url"]) as response:
            with target_path.open("wb") as output_file:
                copyfileobj(response, output_file)

        download_results.append({
            "file_name": candidate["file_name"],
            "status": "downloaded",
            "size_bytes": target_path.stat().st_size,
        })

    except Exception as error:
        download_results.append({
            "file_name": candidate["file_name"],
            "status": "error",
            "error": str(error),
        })

download_results
```

```python
downloaded_files = sorted(
    path for path in REFERENCE_DIR.iterdir()
    if path.is_file() and path.suffix.lower() == ".mp4"
)

print(f"Videos locales encontrados: {len(downloaded_files)}")

for path in downloaded_files:
    print(f"- {path.name}: {path.stat().st_size / (1024 ** 2):.2f} MB")
```

```python
import cv2

video_inventory = []

for video_path in downloaded_files:
    capture = cv2.VideoCapture(str(video_path))

    opened = capture.isOpened()
    frame_count = int(capture.get(cv2.CAP_PROP_FRAME_COUNT))
    fps = capture.get(cv2.CAP_PROP_FPS)
    width = int(capture.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(capture.get(cv2.CAP_PROP_FRAME_HEIGHT))

    duration_sec = frame_count / fps if fps > 0 else None

    capture.release()

    video_inventory.append({
        "file_name": video_path.name,
        "file_size_bytes": video_path.stat().st_size,
        "opened": opened,
        "frame_count": frame_count,
        "fps": round(fps, 3),
        "width": width,
        "height": height,
        "duration_sec": round(duration_sec, 3) if duration_sec else None,
    })

for video in video_inventory:
    print(video)
```

## Inventario técnico persistente

La inspección técnica se combina con los metadatos publicados por AIST y
se guarda como un inventario pequeño y reproducible. Este archivo será la
entrada declarativa del Notebook 02.

```python
import csv

inventory_by_name = {
    video["file_name"]: video
    for video in video_inventory
}

reference_video_inventory = []
for candidate in candidates_by_genre.values():
    technical = inventory_by_name[candidate["file_name"]]
    reference_video_inventory.append({
        **candidate,
        "opened": technical["opened"],
        "frame_count": technical["frame_count"],
        "fps": technical["fps"],
        "width": technical["width"],
        "height": technical["height"],
    })

inventory_fields = [
    "file_name", "url", "genre", "situation",
    "camera", "dancer", "duration_sec", "file_size_bytes",
    "opened", "frame_count", "fps", "width", "height",
]

inventory_path = METADATA_DIR / "reference_video_inventory.csv"
with inventory_path.open("w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(file, fieldnames=inventory_fields)
    writer.writeheader()
    writer.writerows(reference_video_inventory)

print(f"Inventario guardado en: {inventory_path}")
print(f"Registros del inventario: {len(reference_video_inventory)}")
```
